```python
from pydub import AudioSegment 
```

```python
from pydub.generators import WhiteNoise
```

```python
import os
```

```python
import random
```

```python
# input_audio_path (str): Path to the user's input audio file.
# output_audio_path (str): Path where the output audio with noise will be saved.
# duration (int): Duration in milliseconds for the final audio length.
# noise_level (float): Volume of the background noise in decibels (default: -20 dB).
```

```python
input_audio_path = "/mnt/c/Users/lynet/OneDrive/Documents/github_code_conap/conap/add_white_noise/test_white_noise/BAK.wav"
```

```python
output_audio_path = "/mnt/c/Users/lynet/OneDrive/Documents/github_code_conap/conap/add_white_noise/test_white_noise/output_with_noise.wav"
```

```python
duration = 500
```

```python
noise_level = -20.0
```

```python
# Load user audio
# AudioSegment manipulates audio files (e.g. loading and slicing audio)
user_audio = AudioSegment.from_file(input_audio_path)
```

```python
# Check if length of input audio can support the indicated duration
if len(user_audio) < duration:
    raise ValueError (f"Input audio is only {len(user_audio)} ms long, but requested {duration} ms.")
```

```python
# Choose a random start point for the audio clip
max_start = len(user_audio) - duration
start_point = random.randint(0, max_start)
user_audio_segment = user_audio[start_point:start_point + duration]
```

```python
max_start
```

```python
start_point
```

```python
user_audio_segment
```

```python
# Generate white noise
noise = WhiteNoise().to_audio_segment(duration=duration).apply_gain(noise_level)
```

```python
noise
```

```python
# Mix audio and noise
mixed = user_audio_segment.overlay(noise)
```

```python
mixed
```

```python
# Export the result
mixed.export(output_audio_path, format="wav")
print(f"Output saved to: {output_audio_path}")
```
