```python
from pydub import AudioSegment
```

```python
input_path = "/mnt/c/Users/lynet/OneDrive/Documents/github_code_conap/conap/change_volume/test_change_volume/BAK.wav"
```

```python
output_path = "/mnt/c/Users/lynet/OneDrive/Documents/github_code_conap/conap/change_volume/test_change_volume/change_volume_audio.wav"
```

```python
gain_db = +20
```

```python
# Load the audio file
audio = AudioSegment.from_file(input_path)
```

```python
audio
```

```python
# Apply volume change
adjusted_audio = audio.apply_gain(gain_db)
```

```python
adjusted_audio
```

```python
# Export the modified audio
adjusted_audio.export(output_path, format="wav")
print(f"Volume adjusted by {gain_db} dB and saved to: {output_path}")
```
