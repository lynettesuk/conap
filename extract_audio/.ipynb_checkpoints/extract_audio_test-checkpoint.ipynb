```python
from pydub import AudioSegment
```

```python
import random
```

```python
input_audio = "/mnt/c/Users/lynet/OneDrive/Documents/github_code_conap/conap/extract_audio/test_audio/BAK.wav"
```

```python
clip_length = 500
```

```python
audio = AudioSegment.from_wav(input_audio)
   
```

```python
type(audio)
```

```python
#Duration of the audio in milliseconds
duration = len(audio)    
```

```python
duration
```

```python

# Duration of the clip you want to extract (in ms)
clip_duration = clip_length
```

```python
# Make sure the clip fits in the audio
if duration < clip_duration:
    raise ValueError(f"Audio file is too short for a {clip_duration}-millisecond clip.")
```

```python
# Choose a random start point
start = random.randint(0, duration - clip_duration)
end = start + clip_duration
```

```python
start
```

```python
end
```

```python
# Slice the audio
clip = audio[start:end]
```

```python
clip
```

```python
# Export the extracted clip
exported_clip= clip.export("clip.wav", format="wav")
print(f"Extracted 0.5-second clip from {start}ms to {end}ms.")
```
